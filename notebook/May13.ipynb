# Fix the checkpoints for single GPU inference

```python
import os
import torch

def remove_module_prefix_from_checkpoints(directory):
    """
    Iterate through all .pth files in the given directory and remove the 'module.' prefix
    from the keys in model_state_dict, then overwrite the checkpoint file.
    """
    for filename in os.listdir(directory):
        if filename.endswith(".pt"):
            filepath = os.path.join(directory, filename)
            print(f"Processing {filepath}...")

            checkpoint = torch.load(filepath, map_location='cpu', weights_only=False)

            if "model_state_dict" not in checkpoint:
                print(f"⚠️  Skipping {filename}: No 'model_state_dict' found.")
                continue

            state_dict = checkpoint["model_state_dict"]
            new_state_dict = {}

            for key, value in state_dict.items():
                if key.startswith("module."):
                    new_key = key[7:]  # strip 'module.'
                else:
                    new_key = key
                new_state_dict[new_key] = value

            checkpoint["model_state_dict"] = new_state_dict

            # Overwrite the file with the updated state dict
            torch.save(checkpoint, filepath)
            print(f"✅ Updated and saved {filename}\n")

# Example usage:
remove_module_prefix_from_checkpoints("/home/yb107/cvit-work/cvpr2025/SAT/checkpoints")
```
